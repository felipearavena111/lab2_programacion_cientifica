```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración de estilo para los gráficos 
sns.set_theme(style="whitegrid")
plt.rcParams['figure.dpi'] = 120  
plt.rcParams['font.size'] = 11
```

```python
# Carga y exploracion del dataset
df = pd.read_csv('Iris.csv')
df = df[['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm', 'Species']]
```

Se carga el dataset y se reescribe para no trabajar con la columna ID, ya que solamente identifica cada registro y no aporta información para el análisis

```python
print(f"Dimensiones del dataset: {df.shape}")

print(f"\nComprobar valores nulos:\n{df.isnull().sum()}")

print(f"\nCantidad de registros por especie:\n {df['Species'].value_counts()}")

print(f"\nEstadisticas descriptivas: \n{df.describe()}")

print(f"\nInformacion estructural del dataset:\n")
df.info()
```

Se analiza el dataset, se revisan dimensiones, valores nulos, los tipos de datos, la cantidad de muestras y se calculan las estadísticas descriptivas básicas.

```python
# Obtener estadisticas descriptivas por cada especie
print("Resumen Estadístico por Especie")
estadisticas_por_especie = df.groupby("Species").describe().T
display(estadisticas_por_especie)

# Analisis simplificado: Promedios de las dimensiones por especie
print("\nPromedios Morfológicos por Especie")
promedios = df.groupby("Species").mean()
display(promedios)
```

Se hace un resumen estadístico por cada especie para tener una idea general de los valores y un promedio general de las mediciones de cada especie.

```python
# Detección visual de Valores Atípicos 

# Creacion de lienzo
plt.figure(figsize=(12, 8))

# Grafico 1: Longitud del Sépalo
plt.subplot(2, 2, 1)  # (2 filas, 2 columnas, posición 1)
sns.boxplot(x='Species', y='SepalLengthCm', hue='Species', data=df, palette='pastel', legend=False, boxprops=dict(alpha=0.9))
plt.title('Distribución de: Longitud del Sépalo', fontweight='bold', fontsize=12)
plt.xlabel('Especie de Iris', fontsize=10)
plt.ylabel('Centímetros (cm)', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.4)


# Grafico 2: Ancho del Sépalo
plt.subplot(2, 2, 2)  # (2 filas, 2 columnas, posición 2)
sns.boxplot(x='Species', y='SepalWidthCm', hue='Species', data=df, palette='pastel', legend=False, boxprops=dict(alpha=0.9))
plt.title('Distribución de: Ancho del Sépalo', fontweight='bold', fontsize=12)
plt.xlabel('Especie de Iris', fontsize=10)
plt.ylabel('Centímetros (cm)', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.4)


# Grafico 3: Longitud del Pétalo
plt.subplot(2, 2, 3)  # (2 filas, 2 columnas, posición 3)
sns.boxplot(x='Species', y='PetalLengthCm', hue='Species', data=df, palette='pastel', legend=False, boxprops=dict(alpha=0.9))
plt.title('Distribución de: Longitud del Pétalo', fontweight='bold', fontsize=12)
plt.xlabel('Especie de Iris', fontsize=10)
plt.ylabel('Centímetros (cm)', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.4)


# Grafico 4: Ancho del Pétalo
plt.subplot(2, 2, 4)  # (2 filas, 2 columnas, posición 4)
sns.boxplot(x='Species', y='PetalWidthCm', hue='Species', data=df, palette='pastel', legend=False, boxprops=dict(alpha=0.9))
plt.title('Distribución de: Ancho del Pétalo', fontweight='bold', fontsize=12)
plt.xlabel('Especie de Iris', fontsize=10)
plt.ylabel('Centímetros (cm)', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.4)

# Ajustes de guardado
plt.tight_layout()
plt.savefig('boxplot_atipicos.png', dpi=150, bbox_inches='tight')
plt.show()
```

Se encuentran algunos valores atípicos, decidí dejarlos porque al ser datos de flores reales, es normal que la naturaleza tenga pequeñas variaciones, así que no son considerados errores de medición.

```python
# Pregunta 1: Diferencias en longitud de sépalo

resumen_sepalo = df.groupby("Species")["SepalLengthCm"].describe()
display(resumen_sepalo)
```

Para la primera pregunta se quería comparar la longitud del sépalo de todas las especies. Los datos obtenidos nos dice que la longitud de sepálo de la especie setosa es menor a comparación a las demas especies.

```python
# Pregunta 2: Correlacion entre longitud y ancho del petalo

correlacion = df["PetalLengthCm"].corr(df["PetalWidthCm"])
print(f"El nivel de correlación matemática es de: {correlacion:.4f}")

# Configurar el tamaño del gráfico
plt.figure(figsize=(8, 5))

# Separamos los datos de cada especie 
setosa = df[df['Species'] == 'Iris-setosa']
versicolor = df[df['Species'] == 'Iris-versicolor']
virginica = df[df['Species'] == 'Iris-virginica']

# Graficamos cada especie 
plt.scatter(x=setosa["PetalLengthCm"], y=setosa["PetalWidthCm"], 
            c='skyblue', label='Setosa', s=100, alpha=0.8)
plt.scatter(x=versicolor["PetalLengthCm"], y=versicolor["PetalWidthCm"], 
            c='orange', label='Versicolor', s=100, alpha=0.8)
plt.scatter(x=virginica["PetalLengthCm"], y=virginica["PetalWidthCm"], 
            c='lightgreen', label='Virginica', s=100, alpha=0.8)

# Personalización del gráfico
plt.title("Relación: Longitud vs Ancho del Pétalo", fontweight='bold', fontsize=12)
plt.xlabel("Longitud del Pétalo (cm)", fontsize=10)
plt.ylabel("Ancho del Pétalo (cm)", fontsize=10)
plt.legend(title="Especie")
plt.grid(True, linestyle='--', alpha=0.4)

# Ajustes de guardado
plt.tight_layout()
plt.savefig('p2_petalo_scatter.png', dpi=150)
plt.show()
```

Se quiere comprobar la relación entre el ancho y largo del pétalo, se demuestra que ambas medidas crecen juntas y de manera proporcional en las tres especies.

```python
# Pregunta 3: Diferenciar especies por las medidas
pair_plot = sns.pairplot(
    df, 
    hue="Species", 
    palette="pastel", 
    diag_kind="kde",
    plot_kws={'alpha': 0.8, 's': 50}
)


pair_plot.fig.suptitle("Matriz de Relaciones Morfológicas por Especie",
                       y=1.02, fontweight='bold', fontsize=14)

pair_plot.savefig('p3_matriz_diferenciacion.png', dpi=150, bbox_inches='tight')
plt.show()
```

Para la última pregunta se generan múltiples gráficos cruzados. Se descubre que al analizar los gráficos del pétalo , las tres especies forman grupos independientes. En cambio, cuando cruzamos las medidas del sépalo, los valores se mezclan en gran medida, por lo que se concluye que para diferenciar las especies son más útiles los valores del pétalo.
